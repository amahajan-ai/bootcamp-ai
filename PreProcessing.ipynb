```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Sample data generation
n_students = 100  # Number of students

data = {
    "StudentID": np.arange(1, n_students + 1),
    "G1": np.random.randint(0, 20, n_students),
    "G2": np.random.randint(0, 20, n_students),
    "G3": np.random.randint(0, 20, n_students),
    "Mjob": np.random.choice(['teacher', 'health', 'services', 'at_home', 'other'], n_students),
    "Fjob": np.random.choice(['teacher', 'health', 'services', 'at_home', 'other'], n_students),
    "age": np.random.randint(15, 22, n_students),
    "address": np.random.choice(['urban', 'rural'], n_students),
    "studytime": np.random.randint(1, 5, n_students),
    "failures": np.random.randint(0, 4, n_students),
    "schoolsup": np.random.choice(['yes', 'no'], n_students),
    "famsup": np.random.choice(['yes', 'no'], n_students),
    "paid": np.random.choice(['yes', 'no'], n_students),
    "activities": np.random.choice(['yes', 'no'], n_students),
    "nursery": np.random.choice(['yes', 'no'], n_students),
    "higher": np.random.choice(['yes', 'no'], n_students),
    "internet": np.random.choice(['yes', 'no'], n_students),
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the dataset
print(df.head())
#df.to_csv('student_performance.csv', index=False)
```

```python
### Step 1: Load the Dataset and Handle Missing Data
import pandas as pd

# Load dataset (assuming a CSV file named 'student_performance.csv')
df = pd.read_csv('student_performance.csv')
```

```python
# Check for missing data
print(df.isnull().sum())

# Strategies to handle missing data:
# 1. Replace missing data with mean, median, or mode.
# 2. Remove rows with missing data.

# Example: Replace missing numerical data with the mean
df.fillna(df.mean(), inplace=True)
```

```python
# Example: For categorical data, replace missing data with the mode
for column in df.select_dtypes(include=['object']).columns:
    df.fillna(df[column].mode()[0], inplace=True)
```

1. df.select_dtypes(include=['object']).columns:
df.select_dtypes(): This method is used to select columns of a specific data type.
include=['object']: This specifies that only columns with the data type 'object' (typically used for strings or categorical data) should be selected.
.columns: This extracts the column names of the selected columns.
This part of the code identifies all the columns in the DataFrame df that have a data type of 'object'.

2. for column in df.select_dtypes(include=['object']).columns::
This starts a for loop to iterate over each column name identified as an 'object' type.
3. df[column].fillna(df[column].mode()[0], inplace=True):
df[column]: Refers to the current column in the loop.
fillna(): This method fills missing (NaN) values in the column.
df[column].mode(): Calculates the mode (the most frequently occurring value) of the current column.
df[column].mode()[0]: Since mode() returns a Series, [0] is used to access the first (and typically the only) mode value.
inplace=True: This modifies the DataFrame in place, meaning the original DataFrame df is updated directly without needing to assign the result to a new variable

```python
### Step 2: Display Statistical Information and Data Types: Display basic statistical information and check data types. 

# Display statistical information
print(df.describe())
```

```python
# Display data types
print(df.dtypes)
```

```python
### Step 3: Create a New Column `average_score`

### Create a new column that calculates the mean of the three exam scores: G1, G2, and G3.

# Ensure G1, G2, G3 are numerical
df[['G1', 'G2', 'G3']] = df[['G1', 'G2', 'G3']].apply(pd.to_numeric, errors='coerce')

# Create the average_score column
df['average_score'] = df[['G1', 'G2', 'G3']].mean(axis=1)

print(df[['G1', 'G2', 'G3', 'average_score']].head())
```

df[['G1', 'G2', 'G3']]:

This selects the columns G1, G2, and G3 from the DataFrame df.
The selection returns a new DataFrame consisting of only these columns.

apply(pd.to_numeric):
The apply method applies a function along an axis of the DataFrame.
pd.to_numeric is a Pandas function that attempts to convert its input to a numeric data type (integer or float).
This function is applied to each element in the selected columns.

errors='coerce':
The errors parameter specifies how to handle errors during the conversion.
'coerce' means that any value that cannot be converted to a numeric type (e.g., a string that doesn't represent a number) will be replaced with NaN (Not a Number).

df[['G1', 'G2', 'G3']] =:
This reassigns the converted columns back to the original DataFrame df.
It updates the G1, G2, and G3 columns with their new, numeric versions.

Purpose

You have columns with mixed types (e.g., numbers stored as strings).
You want to ensure that these columns are of numeric types for calculations.
You handle any non-numeric data gracefully by converting it to NaN.

```python
### Step 4: Find Students Scoring Above 5 in G3 and Count by Mother's Job

###Find students who scored above 5 in G3 and display the count by their mother's job.

# Filter students with G3 > 5
above_five_g3 = df[df['G3'] > 5]
print(above_five_g3)
# The code above_five_g3 = df[df['G3'] > 5] filters a DataFrame df to create a new DataFrame above_five_g3 containing
# only the rows where the value in column G3 is greater than 5. 

# Count by Mother's job (Mjob)
count_by_mjob = above_five_g3['Mjob'].value_counts()

print(count_by_mjob)
```
